```python
def linear_search(l1,target) :
  for i in range(len(l1)) :
    if l1[i] == target :
      return i
    return -1
l1 = [int(x) for x in input("enter the list").split()]
target = int(input("enter the target :"))
i = linear_search(l1,target)
if i == -1 :
  print("not found")
else :
  print(l1[i],"is at index",linear_search(l1,target))
```

Binary search

```python
def binary_search(l1, target):
    left, right = 0, len(l1) - 1
    while left <= right:
        mid = (left + right) // 2
        if l1[mid] == target:
            return mid  # Target found at index mid
        elif l1[mid] < target:
            left = mid + 1  # Search right half
        else:
            right = mid - 1  # Search left half
    return -1
l1 = [int(x) for x in input("enter the list").split()]
target = int(input("enter the target :"))
result = binary_search(l1, target)
print("Target found at index:", result)
```
